```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv(r'D:\kathan\python_codes\7th sem\ML lab\ML-lab\dataset\Mall_Customers_cluster.csv')
data.head()
```

```python
x = data.iloc[:,3:5].values  # taking all values of last 2 columns
```

```python
x
```

```python
from sklearn.cluster import KMeans
n=[]

for i in range(1, 11): ## clusters value can never be 0 so always start with 1
    kmeans = KMeans(n_clusters=i, random_state=0)
    kmeans.fit(x)
    n.append(kmeans.inertia_)
    
```

```python
plt.plot(range(1,11),n)
plt.xlabel("no of clusters")
plt.ylabel("WCSS")
plt.title("Elow method")
```

```python
model = KMeans(n_clusters=5, random_state=0)
Y=model.fit_predict(x)
```

```python
from sklearn.metrics import silhouette_score
print(silhouette_score(x, Y))
```
